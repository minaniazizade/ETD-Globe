```python
import xarray as xr
import numpy as np
import pandas as pd
import glob
import math

files = sorted(glob.glob("Dataset/*.nc4"))

print("Number of files:", len(files))

ds = xr.open_mfdataset(
    files,
    combine="by_coords",
    engine="netcdf4"
)

print(ds)


dims = list(ds.sizes.keys())
sizes = [ds.sizes[d] for d in dims]

total_rows = math.prod(sizes)

print("Total possible rows:", total_rows)

n = min(500000, total_rows)

rng = np.random.default_rng(42)

random_indices = rng.choice(
    total_rows,
    size=n,
    replace=False
)

indices = np.unravel_index(random_indices, sizes)

indexers = {
    dim: xr.DataArray(indices[i], dims="sample")
    for i, dim in enumerate(dims)
}

sampled_ds = ds.isel(indexers)

df = sampled_ds.to_dataframe().reset_index()


df.to_csv("Dataset.csv", index=False)

print("Final shape:", df.shape)
print("CSV saved successfully!")

ds.close()
```
